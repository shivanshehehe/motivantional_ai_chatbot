```python

from openai import OpenAI
from datetime import datetime

client = OpenAI(
    api_key="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"),

chat_history = []


def detect_mood(user_input):

    text = user_input.lower()

    if "stress" in text or "nervous" in text or "anxious" in text:
        return "stressed"

    elif "sad" in text or "upset" in text:
        return "sad"

    elif "happy" in text or "excited" in text:
        return "happy"

    elif "tired" in text or "exhausted" in text:
        return "tired"

    else:
        return "normal"

def generate_response(user_input, mood):

    prompt = f"""
    You are a supportive AI Study and Motivation Assistant.

    User mood: {mood}

    User says:
    {user_input}

    Your task:
    1. Understand the user's emotion
    2. Give motivation
    3. Give practical study advice
    4. Keep response simple and positive
    5. Keep response under 120 words
    """

    response = client.chat.completions.create(
        model="gpt-4.1-mini",

        messages=[
            {
                "role": "system",
                "content": "You are a helpful AI mentor for students."
            },

            {
                "role": "user",
                "content": prompt
            }
        ]
    )

    return response.choices[0].message.content


def save_chat(user, ai):

    with open("chat_history.txt", "a", encoding="utf-8") as file:

        file.write(f"\nTime: {datetime.now()}\n")
        file.write(f"User: {user}\n")
        file.write(f"AI: {ai}\n")
        file.write("-" * 50)



def show_menu():

    print("\n========== AI STUDY & MOTIVATION ASSISTANT ==========")
    print("1. Chat with AI")
    print("2. View Chat History")
    print("3. Exit")
    print("=====================================================")

def view_history():

    if len(chat_history) == 0:
        print("\nNo chat history available.\n")

    else:
        print("\n========== CHAT HISTORY ==========\n")

        for chat in chat_history:

            print("You:", chat["user"])
            print("AI :", chat["ai"])
            print()


print("\nWelcome to AI Study & Motivation Assistant!")

while True:

    show_menu()

    choice = input("\nEnter your choice: ")


    if choice == "1":

        user_input = input("\nTell me your problem, mood, or study goal:\n")

        # Detect mood
        mood = detect_mood(user_input)

        print("\nDetected Mood:", mood)

        # Generate AI response
        ai_response = generate_response(user_input, mood)

        print("\n========== AI RESPONSE ==========\n")
        print(ai_response)

        # Save in session memory
        chat_history.append({
            "user": user_input,
            "ai": ai_response
        })

        # Save in text file
        save_chat(user_input, ai_response)



    elif choice == "2":

        view_history()

   

    elif choice == "3":

        print("\nThank you for using the AI Assistant!")
        print("Keep learning and stay confident!\n")

        break

   
    else:

        print("\nInvalid choice. Please try again.\n")
```
